# Lecture 15 · Subword-level Self-Attention 이해

- **문자 단위(char-level)가 아니라 서브워드 단위(subword-level)** 실습입니다. (문자 단위 텍스트 생성은 Lecture 14 참고)
- 토크나이저: `klue/bert-base` (한국어 subword tokenizer)
- 목표: Self-Attention이 Query·Key·Value로부터 가중치를 어떻게 계산하는지 직접 관찰

> **주의(중요)**: 아래 예시는 문장 30개 수준의 아주 작은 장난감(toy) 데이터입니다.
> 학습 후 "그것"이 명사를 더 잘 참조하도록 바뀐 것처럼 보이더라도, 이는 **모델이 대명사 참조(coreference)를
> "이해"했다는 증거가 아니라 Self-Attention 메커니즘 자체(패딩 마스크, causal mask, 위치 정보가 가중치 계산에
> 미치는 영향)를 관찰하기 위한 예시**로 해석해야 합니다.


```python
# --- 환경 설정 ---
import sys

if "google.colab" in sys.modules:
    import subprocess
    subprocess.run(["pip", "install", "-q", "koreanize-matplotlib", "transformers"], check=True)

import numpy as np, tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from transformers import AutoTokenizer
import matplotlib.pyplot as plt
import koreanize_matplotlib

SEED = 42;  np.random.seed(SEED);  tf.random.set_seed(SEED)

# ---- 설정 ----
D_MODEL = 16
N_HEADS = 1
EPOCHS  = 40
BATCH   = 64

# ---- 실험 권장 (1~2개만 수정) ----
# D_MODEL = 32   # 차원 증가    -> Attention 패턴이 더 뚜렷해지는가?
# N_HEADS = 2    # 헤드 수 증가 -> 참조 패턴이 다양해지는가?
# EPOCHS  = 80   # 에폭 증가    -> 학습 전/후 차이가 더 커지는가?
```

## 1. 데이터

기존 버전은 문장 6개를 100번 복제한 것이라 다양성이 없었습니다. 아래에서는 주어·장소·이유를
다양화한 문장 **30개**를 준비합니다. 모두 "그것이 ~해서" 구조로 끝나며, "그것"은 바로 앞에 등장한
장소/사물 명사를 가리키도록 설계했습니다.

```python
# --- 데이터 (30개 고유 문장, 다양한 주어/장소/이유) ---
sentences = [
    "고양이가 매트 위에 앉았다 그것이 부드러워서",
    "고양이가 소파에 누웠다 그것이 따뜻해서",
    "강아지가 공원으로 달려갔다 그것이 넓어서",
    "강아지가 마당에서 뛰어놀았다 그것이 즐거워서",
    "새가 나무로 날아갔다 그것이 높아서",
    "새가 하늘을 날았다 그것이 시원해서",
    "토끼가 풀밭에서 뛰었다 그것이 푹신해서",
    "토끼가 굴로 들어갔다 그것이 아늑해서",
    "다람쥐가 나무를 올랐다 그것이 튼튼해서",
    "다람쥐가 도토리를 모았다 그것이 맛있어서",
    "곰이 강으로 걸어갔다 그것이 맑아서",
    "곰이 동굴에서 잠들었다 그것이 따뜻해서",
    "여우가 숲으로 들어갔다 그것이 조용해서",
    "여우가 언덕을 넘었다 그것이 가팔라서",
    "사슴이 초원에서 뛰었다 그것이 드넓어서",
    "사슴이 시냇물을 건넜다 그것이 시원해서",
    "펭귄이 얼음판에서 미끄러졌다 그것이 매끄러워서",
    "펭귄이 바다로 뛰어들었다 그것이 차가워서",
    "코끼리가 호수로 들어갔다 그것이 깊어서",
    "코끼리가 진흙을 밟았다 그것이 부드러워서",
    "원숭이가 나뭇가지를 잡았다 그것이 단단해서",
    "원숭이가 바나나를 먹었다 그것이 달콤해서",
    "판다가 대나무를 씹었다 그것이 신선해서",
    "판다가 언덕에서 굴렀다 그것이 재미있어서",
    "고래가 바다를 헤엄쳤다 그것이 광활해서",
    "고래가 물살을 갈랐다 그것이 세차서",
    "부엉이가 나뭇가지에 앉았다 그것이 튼튼해서",
    "부엉이가 밤하늘을 날았다 그것이 고요해서",
    "거북이가 모래밭을 기어갔다 그것이 따뜻해서",
    "거북이가 바위 위에 올랐다 그것이 널찍해서",
] * 20
print(f"총 문장 수: {len(sentences)}  (고유 문장 30개 x 20회 반복)")
```

## 2. 토크나이징 & 입력 구성

- `pad_id == 0` 이므로 `Embedding(mask_zero=True)`로 패딩을 자동으로 무시합니다.
- 위치 정보가 없으면 Self-Attention은 순서를 전혀 알 수 없으므로, **위치 임베딩(position embedding)**을
  토큰 임베딩에 더해줍니다.
- 다음 토큰 예측(next-token prediction) 목적이므로, 각 위치는 **자신보다 앞의 토큰만** 볼 수 있어야
  합니다 (`causal mask`). 이는 모델 정의 단계에서 `use_causal_mask=True`로 적용합니다.

```python
# --- 토크나이저 & 인코딩 ---
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")
SEQ_LEN   = max(len(tokenizer.tokenize(s)) for s in sentences[:30]) + 1
VOCAB     = tokenizer.vocab_size
pad_id    = tokenizer.pad_token_id
assert pad_id == 0, "이 실습은 pad_id==0을 전제로 mask_zero=True를 사용합니다."

def encode_m2m(sents):
    X, Y = [], []
    for s in sents:
        ids = tokenizer.encode(s, add_special_tokens=False)
        ids = (ids + [pad_id] * (SEQ_LEN + 1))[:SEQ_LEN + 1]
        X.append(ids[:-1]);  Y.append(ids[1:])
    return np.array(X, dtype=np.int32), np.array(Y, dtype=np.int32)

X, y = encode_m2m(sentences)
pos_ids = np.tile(np.arange(SEQ_LEN, dtype=np.int32), (len(X), 1))

dataset = tf.data.Dataset.from_tensor_slices(((X, pos_ids), y)).batch(BATCH)
print(f"SEQ_LEN={SEQ_LEN}  X={X.shape}  VOCAB={VOCAB:,}")
```

## 3. 모델 (Functional API)

`keras.Model`을 상속하는 커스텀 클래스 대신, 초보자에게 더 친숙한 **Functional API**로 구성합니다.
학습용 모델(`train_model`)과 Attention 가중치까지 함께 반환하는 시각화용 모델(`attn_model`)은
**같은 레이어(가중치 공유)**를 사용하므로 둘 중 하나만 학습시켜도 됩니다.

> 패딩 마스크는 토큰 임베딩에 위치 임베딩을 더하는 과정에서 유실될 수 있으므로,
> `keras.ops.not_equal(token_input, 0)`로 직접 만들어 `attention_mask`로 전달합니다
> (Lab3의 Conv1D 마스킹과 동일한 방식이며, 커스텀 `Lambda` 레이어를 쓰지 않습니다).

```python
# --- 모델 ---
def build_self_attention_lm(vocab_size, seq_len, d_model, n_heads):
    token_input = layers.Input(shape=(seq_len,), dtype="int32", name="tokens")
    pos_input   = layers.Input(shape=(seq_len,), dtype="int32", name="positions")

    tok_emb = layers.Embedding(vocab_size, d_model, name="token_emb")(token_input)
    pos_emb = layers.Embedding(seq_len, d_model, name="pos_emb")(pos_input)
    x = tok_emb + pos_emb   # 토큰 임베딩 + 위치 임베딩

    # 패딩 마스크: pad_id(=0) 위치는 key/value로 사용하지 않음 -> (batch, 1, seq_len)
    pad_mask = keras.ops.not_equal(token_input, 0)[:, tf.newaxis, :]

    mha = layers.MultiHeadAttention(num_heads=n_heads, key_dim=d_model // n_heads,
                                     name="self_attention")
    attn_out, attn_scores = mha(query=x, value=x, key=x,
                                 attention_mask=pad_mask,       # 패딩 마스크
                                 use_causal_mask=True,          # 미래 토큰 차단
                                 return_attention_scores=True)

    x = layers.LayerNormalization()(x + attn_out)
    logits = layers.Dense(vocab_size, name="lm_head")(x)

    train_model = keras.Model([token_input, pos_input], logits, name="self_attn_lm")
    attn_model  = keras.Model([token_input, pos_input], [logits, attn_scores],
                              name="self_attn_lm_with_attention")
    return train_model, attn_model

train_model, attn_model = build_self_attention_lm(VOCAB, SEQ_LEN, D_MODEL, N_HEADS)
train_model.compile(loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                     optimizer="adam")
train_model.summary()
```

```python
# --- 유틸 ---
def get_tokens(sentence):
    ids     = tokenizer.encode(sentence, add_special_tokens=False)
    ids     = (ids + [pad_id] * SEQ_LEN)[:SEQ_LEN]
    labels  = tokenizer.convert_ids_to_tokens(ids)
    pos     = np.arange(SEQ_LEN, dtype=np.int32)
    return [np.array([ids], dtype=np.int32), np.array([pos])], labels

sample, token_labels = get_tokens(sentences[0])

# --- 학습 전 Attention ---
_, attn_before = attn_model(sample)
```

```python
# --- 학습 ---
history = train_model.fit(dataset, epochs=EPOCHS, verbose=1)
```

```python
# --- 학습 후 Attention ---
_, attn_after = attn_model(sample)
```

## 4. 시각화

```python
# --- 시각화 1 : Attention 히트맵 전/후 ---
n   = len(token_labels)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
for ax, attn, cmap, title in [
    (ax1, attn_before, "Blues", "학습 전"),
    (ax2, attn_after,  "Reds",  "학습 후"),
]:
    im = ax.imshow(attn[0, 0, :n, :n].numpy(), cmap=cmap)
    ax.set_xticks(range(n));  ax.set_xticklabels(token_labels, rotation=45, ha="right", fontsize=8)
    ax.set_yticks(range(n));  ax.set_yticklabels(token_labels, fontsize=8)
    ax.set_title(title, fontsize=12, fontweight="bold")
    fig.colorbar(im, ax=ax, shrink=0.8)
plt.suptitle("Self-Attention 히트맵 비교 (causal mask 적용: 아래쪽 삼각형만 값을 가짐)",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig("fig06_heatmap.png", dpi=150, bbox_inches="tight");  plt.show()
```

```python
# --- 시각화 2 : "그것" 참조 대상 전/후 ---
noun_candidates = ["매트", "소파", "공원", "마당", "나무", "하늘", "풀밭", "굴", "강", "동굴",
                    "숲", "언덕", "초원", "시냇물", "얼음판", "바다", "호수", "진흙", "나뭇가지",
                    "바나나", "대나무", "물살", "밤하늘", "모래밭", "바위"]
tgt_idx = next(i for i, lb in enumerate(token_labels) if "그것" in lb)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))
for ax, attn, title in [
    (ax1, attn_before, "학습 전 — '그것' 참조"),
    (ax2, attn_after,  "학습 후 — '그것' 참조"),
]:
    scores = attn[0, 0, tgt_idx, :n].numpy()
    colors = ["#E02020" if any(nc in lb for nc in noun_candidates)
              else "#AAAAAA" for lb in token_labels]
    ax.bar(range(n), scores, color=colors)
    ax.set_xticks(range(n));  ax.set_xticklabels(token_labels, rotation=45, ha="right", fontsize=8)
    ax.set_title(title, fontsize=11, fontweight="bold")
    ax.grid(True, axis="y", alpha=0.3)
plt.suptitle("'그것' 참조 대상 (빨강=명사 후보) — causal mask 때문에 '그것' 이후 토큰은 0", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig("fig06_reference.png", dpi=150, bbox_inches="tight");  plt.show()

# --- 결과 ---
for label, attn in [("학습 전", attn_before), ("학습 후", attn_after)]:
    scores = attn[0, 0, tgt_idx, :n].numpy()
    top3   = sorted(zip(token_labels, scores), key=lambda x: -x[1])[:3]
    print(f"{label} 상위 참조: {', '.join(f'{lb}({sc:.3f})' for lb, sc in top3)}")

print()
print("주의: 위 결과는 30개 문장 수준의 작은 장난감 데이터에서 관찰된 패턴입니다.")
print("모델이 대명사 참조(coreference)를 실제로 '이해'했다는 근거로 해석하지 말고,")
print("Self-Attention이 학습을 통해 가중치를 어떻게 재분배하는지 보여주는 예시로만 받아들이세요.")
```
