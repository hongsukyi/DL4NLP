# Lecture 14 실습 · Character-level LSTM 텍스트 생성

Shakespeare 텍스트를 **문자 단위**로 학습하고 다음 문자를 반복해서 생성합니다.

- 입력: 연속된 `SEQ_LEN`개의 문자
- 정답: 입력 바로 다음 문자
- 모델: `Embedding → LSTM → Dense`

## 1. 라이브러리 준비

```python
!pip install -q koreanize-matplotlib
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import koreanize_matplotlib

from tensorflow import keras
from tensorflow.keras import layers

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

## 2. 실습 설정

빠른 실행을 위한 기본값입니다. 자연스러운 문장보다 **문자 생성 과정의 이해**에 초점을 둡니다.

```python
SEQ_LEN = 20       # 한 번에 읽는 문자 수
EMBED_DIM = 32     # 문자 임베딩 크기
LSTM_UNITS = 64    # LSTM 유닛 수
EPOCHS = 10
BATCH_SIZE = 64
MAX_CHARS = 5000   # 학습에 사용할 문자 수

# 생성 품질을 더 확인하려면 아래 값을 사용해 보세요.
# MAX_CHARS = 20000
# LSTM_UNITS = 128
# EPOCHS = 15
```

## 3. Shakespeare 데이터 읽기

```python
path = keras.utils.get_file(
    "shakespeare.txt",
    "https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt"
)

with open(path, encoding="utf-8") as file:
    text = file.read()[:MAX_CHARS]

print("전체 문자 수:", len(text))
print("텍스트 예시:")
print(text[:200])
```

## 4. 문자를 숫자로 바꾸기

신경망은 문자를 직접 처리할 수 없으므로 각 문자에 정수 ID를 부여합니다.

```python
chars = sorted(set(text))
vocab_size = len(chars)

char_to_id = {char: i for i, char in enumerate(chars)}
id_to_char = {i: char for char, i in char_to_id.items()}
encoded = np.array([char_to_id[char] for char in text])

print("고유 문자 수:", vocab_size)
print("고유 문자 예시:", chars[:20])
print("문자 ID 예시:", encoded[:20])
```

## 5. 입력과 정답 만들기

연속된 `SEQ_LEN`개 문자를 입력으로, 바로 다음 문자를 정답으로 사용합니다.

```python
X = np.array([
    encoded[i:i + SEQ_LEN]
    for i in range(len(encoded) - SEQ_LEN)
])

y = np.array([
    encoded[i + SEQ_LEN]
    for i in range(len(encoded) - SEQ_LEN)
])

print("X 크기:", X.shape)
print("y 크기:", y.shape)
```

```python
input_text = "".join(id_to_char[i] for i in X[0])
target_char = id_to_char[y[0]]

print("첫 번째 입력:", repr(input_text))
print("첫 번째 정답:", repr(target_char))
print("입력 문자 ID:", X[0])
print("정답 문자 ID:", y[0])
```

## 6. Character-level LSTM 모델 만들기

- `Embedding`: 문자 ID를 벡터로 변환합니다.
- `LSTM`: 문자 순서를 읽고 문맥을 만듭니다.
- `Dense`: 다음에 올 각 문자의 확률을 계산합니다.

```python
model = keras.Sequential([
    layers.Input(shape=(SEQ_LEN,)),
    layers.Embedding(vocab_size, EMBED_DIM),
    layers.LSTM(LSTM_UNITS),
    layers.Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

## 7. 모델 학습

```python
history = model.fit(
    X, y,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)
```

## 8. 학습 결과 시각화

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history.history["loss"], marker="o")
axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].grid(alpha=0.3)

axes[1].plot(history.history["accuracy"], marker="o")
axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].grid(alpha=0.3)

plt.suptitle("Character-level LSTM 학습 곡선")
plt.tight_layout()
plt.show()
```

## 9. 새로운 문자 생성하기

예측한 문자를 입력 뒤에 붙이고 같은 과정을 반복합니다. `temperature`가 낮으면 안정적으로, 높으면 다양하게 생성됩니다.

```python
def generate_text(seed, length=200, temperature=1.0):
    if temperature <= 0:
        raise ValueError("temperature는 0보다 커야 합니다.")

    result = list(seed)
    sequence = [char_to_id[char] for char in seed[-SEQ_LEN:]]

    for _ in range(length):
        model_input = np.array([sequence[-SEQ_LEN:]])
        probs = model.predict(model_input, verbose=0)[0]

        log_probs = np.log(probs + 1e-10)
        scaled_probs = log_probs / temperature
        scaled_probs = np.exp(scaled_probs)
        scaled_probs = scaled_probs / scaled_probs.sum()

        next_id = np.random.choice(len(scaled_probs), p=scaled_probs)
        result.append(id_to_char[next_id])
        sequence.append(next_id)

    return "".join(result[len(seed):])
```

## 10. Temperature에 따른 결과 비교

```python
seed = text[:SEQ_LEN]
print("시작 문장:", repr(seed))

for temperature in [0.2, 0.5, 1.0, 1.5]:
    generated = generate_text(
        seed,
        length=150,
        temperature=temperature
    )
    print(f"\n[Temperature = {temperature}]")
    print(generated)
```

## 정리

- Character-level 모델은 문장을 문자 ID의 연속으로 처리합니다.
- LSTM은 앞의 문자 문맥을 이용해 다음 문자의 확률을 계산합니다.
- 선택한 문자를 다시 입력에 붙이면 텍스트를 계속 생성할 수 있습니다.
- Temperature는 생성 결과의 안정성과 다양성을 조절합니다.
