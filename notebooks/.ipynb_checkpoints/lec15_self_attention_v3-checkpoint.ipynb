## 1. 라이브러리 준비

```python
!pip install -q transformers koreanize-matplotlib
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import koreanize_matplotlib

from tensorflow import keras
from tensorflow.keras import layers
from transformers import AutoTokenizer

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

## 2. 실습 설정

작은 지역 어휘를 사용하므로 Colab에서 빠르게 학습할 수 있습니다.

```python
D_MODEL = 32       # 임베딩 벡터 크기
N_HEADS = 1       # Attention head 수
EPOCHS = 100
BATCH_SIZE = 8
```

## 3. 작은 문장 데이터 준비

서로 다른 30개 문장만 사용합니다. 같은 문장을 여러 번 복제하지 않습니다.

```python
sentences = [
    "고양이가 매트 위에 앉았다 그것이 부드러워서",
    "고양이가 소파에 누웠다 그것이 따뜻해서",
    "강아지가 공원으로 달려갔다 그것이 넓어서",
    "강아지가 마당에서 뛰어놀았다 그것이 즐거워서",
    "새가 나무로 날아갔다 그것이 높아서",
    "새가 하늘을 날았다 그것이 시원해서",
    "토끼가 풀밭에서 뛰었다 그것이 푹신해서",
    "토끼가 굴로 들어갔다 그것이 아늑해서",
    "다람쥐가 나무를 올랐다 그것이 튼튼해서",
    "다람쥐가 도토리를 모았다 그것이 맛있어서",
    "곰이 강으로 걸어갔다 그것이 맑아서",
    "곰이 동굴에서 잠들었다 그것이 따뜻해서",
    "여우가 숲으로 들어갔다 그것이 조용해서",
    "여우가 언덕을 넘었다 그것이 가팔라서",
    "사슴이 초원에서 뛰었다 그것이 드넓어서",
    "사슴이 시냇물을 건넜다 그것이 시원해서",
    "펭귄이 얼음판에서 미끄러졌다 그것이 매끄러워서",
    "펭귄이 바다로 뛰어들었다 그것이 차가워서",
    "코끼리가 호수로 들어갔다 그것이 깊어서",
    "코끼리가 진흙을 밟았다 그것이 부드러워서",
    "원숭이가 나뭇가지를 잡았다 그것이 단단해서",
    "원숭이가 바나나를 먹었다 그것이 달콤해서",
    "판다가 대나무를 씹었다 그것이 신선해서",
    "판다가 언덕에서 굴렀다 그것이 재미있어서",
    "고래가 바다를 헤엄쳤다 그것이 광활해서",
    "고래가 물살을 갈랐다 그것이 세차서",
    "부엉이가 나뭇가지에 앉았다 그것이 튼튼해서",
    "부엉이가 밤하늘을 날았다 그것이 고요해서",
    "거북이가 모래밭을 기어갔다 그것이 따뜻해서",
    "거북이가 바위 위에 올랐다 그것이 널찍해서",
]

print("고유 문장 수:", len(sentences))
print("첫 번째 문장:", sentences[0])
```

## 4. 문장을 서브워드로 나누기

`klue/bert-base` 토크나이저는 **서브워드 분리 용도**로만 사용합니다. 사전 학습된 BERT 모델이나 임베딩은 사용하지 않습니다.

```python
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")

token_lists = []
for sentence in sentences:
    tokens = tokenizer.tokenize(sentence)
    tokens.append("[EOS]")
    token_lists.append(tokens)

print("원문:", sentences[0])
print("서브워드:", token_lists[0])
```

## 5. 실습용 작은 서브워드 사전 만들기

KLUE-BERT의 전체 어휘 32,000개 대신, 30개 문장에 실제로 등장한 서브워드만 사용합니다.

```python
all_tokens = sorted({
    token
    for tokens in token_lists
    for token in tokens
})

token_to_id = {"[PAD]": 0, "[UNK]": 1}
for token in all_tokens:
    if token not in token_to_id:
        token_to_id[token] = len(token_to_id)

id_to_token = {
    token_id: token
    for token, token_id in token_to_id.items()
}

pad_id = token_to_id["[PAD]"]
unk_id = token_to_id["[UNK]"]
eos_id = token_to_id["[EOS]"]
vocab_size = len(token_to_id)

print("실습 어휘 크기:", vocab_size)
print("어휘 예시:", list(token_to_id.items())[:20])
```

## 6. 입력과 정답 만들기

정답은 입력을 한 칸 왼쪽으로 이동한 다음 서브워드입니다. 짧은 문장의 빈 위치에는 `[PAD]`를 넣습니다.

```python
max_tokens = max(len(tokens) for tokens in token_lists)
SEQ_LEN = max_tokens - 1

X = []
y = []

for tokens in token_lists:
    ids = [token_to_id[token] for token in tokens]
    ids = ids + [pad_id] * (max_tokens - len(ids))

    X.append(ids[:-1])
    y.append(ids[1:])

X = np.array(X, dtype=np.int32)
y = np.array(y, dtype=np.int32)

position_ids = np.tile(
    np.arange(SEQ_LEN, dtype=np.int32),
    (len(X), 1)
)

sample_weight = (y != pad_id).astype("float32")

print("X 크기:", X.shape)
print("y 크기:", y.shape)
print("문장 최대 길이:", max_tokens)
```

```python
input_tokens = [
    id_to_token[token_id]
    for token_id in X[0]
    if token_id != pad_id
]

target_tokens = [
    id_to_token[token_id]
    for token_id in y[0]
    if token_id != pad_id
]

print("입력 X:")
print(input_tokens)
print("\n정답 y:")
print(target_tokens)
```

## 7. Self-Attention 모델 만들기

- 토큰 임베딩과 위치 임베딩을 더합니다.
- padding mask는 `[PAD]` 토큰을 보지 않게 합니다.
- causal mask는 미래 토큰을 보지 않게 합니다.
- 학습 모델과 시각화 모델은 같은 층과 가중치를 공유합니다.

```python
token_input = keras.Input(
    shape=(SEQ_LEN,),
    dtype="int32",
    name="tokens"
)

position_input = keras.Input(
    shape=(SEQ_LEN,),
    dtype="int32",
    name="positions"
)

token_embedding = layers.Embedding(
    vocab_size, D_MODEL, name="token_embedding"
)(token_input)

position_embedding = layers.Embedding(
    SEQ_LEN, D_MODEL, name="position_embedding"
)(position_input)

x = layers.Add()([token_embedding, position_embedding])
```

```python
padding_mask = keras.ops.not_equal(token_input, pad_id)
padding_mask = keras.ops.expand_dims(padding_mask, axis=1)

attention_layer = layers.MultiHeadAttention(
    num_heads=N_HEADS,
    key_dim=D_MODEL // N_HEADS,
    name="self_attention"
)

attention_output, attention_scores = attention_layer(
    query=x,
    value=x,
    key=x,
    attention_mask=padding_mask,
    use_causal_mask=True,
    return_attention_scores=True
)

x = layers.Add()([x, attention_output])
x = layers.LayerNormalization()(x)
logits = layers.Dense(vocab_size, name="next_token")(x)

train_model = keras.Model(
    [token_input, position_input],
    logits,
    name="self_attention_model"
)

attention_model = keras.Model(
    [token_input, position_input],
    [logits, attention_scores],
    name="attention_viewer"
)
```

```python
train_model.compile(
    optimizer="adam",
    loss=keras.losses.SparseCategoricalCrossentropy(
        from_logits=True
    ),
    weighted_metrics=["accuracy"]
)

train_model.summary()
```

## 8. 학습 전 Attention 저장

같은 문장의 학습 전후 Attention을 비교하기 위해 학습 전 값을 먼저 저장합니다.

```python
sample_inputs = [X[:1], position_ids[:1]]
_, attention_before = attention_model(
    sample_inputs,
    training=False
)
attention_before = attention_before.numpy().copy()
```

## 9. 모델 학습

`sample_weight`를 사용하여 `[PAD]` 위치를 loss와 accuracy 계산에서 제외합니다.

```python
history = train_model.fit(
    [X, position_ids],
    y,
    sample_weight=sample_weight,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=0
)

print(f"마지막 loss: {history.history['loss'][-1]:.4f}")
print(f"마지막 accuracy: {history.history['accuracy'][-1]:.4f}")
```

## 10. 학습 곡선 확인

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history.history["loss"])
axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].grid(alpha=0.3)

axes[1].plot(history.history["accuracy"])
axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].grid(alpha=0.3)

plt.suptitle("Self-Attention 모델 학습 곡선")
plt.tight_layout()
plt.show()
```

## 11. 다음 서브워드 예측 확인

각 입력 위치에서 실제 다음 서브워드와 모델이 예측한 서브워드를 비교합니다.

```python
prediction_logits = train_model.predict(
    sample_inputs,
    verbose=0
)
predicted_ids = prediction_logits.argmax(axis=-1)[0]
valid_n = np.count_nonzero(X[0] != pad_id)

print("현재 토큰 → 실제 다음 토큰 / 예측 토큰")
print("-" * 55)

for i in range(valid_n):
    current_token = id_to_token[X[0, i]]
    true_next = id_to_token[y[0, i]]
    predicted_next = id_to_token[predicted_ids[i]]

    print(
        f"{current_token:>10} → "
        f"정답: {true_next:<10} / 예측: {predicted_next}"
    )
```

## 12. 학습 전후 Attention 히트맵

가로축은 참고하는 토큰, 세로축은 정보를 받는 토큰입니다. causal mask 때문에 미래 방향은 0이 되어 아래쪽 삼각형만 나타납니다.

```python
_, attention_after = attention_model(
    sample_inputs,
    training=False
)
attention_after = attention_after.numpy()

valid_labels = input_tokens[:valid_n]
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

items = [
    (attention_before, "Blues", "학습 전"),
    (attention_after, "Reds", "학습 후")
]

for ax, (scores, color, title) in zip(axes, items):
    values = scores[0, 0, :valid_n, :valid_n]
    image = ax.imshow(values, cmap=color, vmin=0, vmax=1)

    ax.set_xticks(range(valid_n))
    ax.set_xticklabels(valid_labels, rotation=45, ha="right")
    ax.set_yticks(range(valid_n))
    ax.set_yticklabels(valid_labels)
    ax.set_title(title)
    ax.set_xlabel("참고하는 토큰")
    ax.set_ylabel("정보를 받는 토큰")
    fig.colorbar(image, ax=ax, shrink=0.8)

plt.suptitle("Self-Attention 가중치: 학습 전과 학습 후")
plt.tight_layout()
plt.show()
```

## 13. 선택한 토큰의 Attention 확인

`그것` 위치가 **이전 토큰들**에 둔 Attention만 확인합니다. 높은 가중치는 정보 결합의 정도를 나타내지만, 대명사 이해의 직접적인 증거는 아닙니다.

```python
target_index = next(
    i
    for i, token in enumerate(valid_labels)
    if "그것" in token
)

previous_tokens = valid_labels[:target_index]
scores_before = attention_before[
    0, 0, target_index, :target_index
]
scores_after = attention_after[
    0, 0, target_index, :target_index
]

x_axis = np.arange(len(previous_tokens))
width = 0.38

plt.figure(figsize=(10, 4))
plt.bar(
    x_axis - width / 2,
    scores_before,
    width,
    label="학습 전"
)
plt.bar(
    x_axis + width / 2,
    scores_after,
    width,
    label="학습 후"
)
plt.xticks(
    x_axis,
    previous_tokens,
    rotation=45,
    ha="right"
)
plt.ylabel("Attention weight")
plt.title("'그것' 위치가 이전 토큰에 둔 Attention")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 14. 서브워드 이어서 생성하기

마지막 위치에서 확률이 가장 높은 서브워드를 선택하고 입력에 다시 붙입니다. 이 과정을 반복하면 서브워드 단위 텍스트 생성이 됩니다.

```python
def generate_subwords(seed_sentence, max_new_tokens=12):
    seed_tokens = tokenizer.tokenize(seed_sentence)
    sequence = [
        token_to_id.get(token, unk_id)
        for token in seed_tokens
    ]
    generated_ids = []

    for _ in range(max_new_tokens):
        context = sequence[-SEQ_LEN:]
        input_length = len(context)
        padded = context + [pad_id] * (
            SEQ_LEN - input_length
        )

        model_x = np.array(
            [padded],
            dtype=np.int32
        )
        model_pos = np.array(
            [np.arange(SEQ_LEN)],
            dtype=np.int32
        )

        prediction = train_model.predict(
            [model_x, model_pos],
            verbose=0
        )

        next_id = int(
            prediction[0, input_length - 1].argmax()
        )

        if next_id == eos_id:
            break

        sequence.append(next_id)
        generated_ids.append(next_id)

    generated_tokens = [
        id_to_token[token_id]
        for token_id in generated_ids
    ]

    return tokenizer.convert_tokens_to_string(
        seed_tokens + generated_tokens
    )
```

```python
seed_sentence = "고양이가 매트 위에"
generated_text = generate_subwords(seed_sentence)

print("시작 문장:", seed_sentence)
print("생성 결과:", generated_text)
```

## 정리

- 서브워드는 문자보다 큰 단위이며 단어보다 작은 조각이 될 수 있습니다.
- 위치 임베딩은 Self-Attention에 토큰의 순서를 알려줍니다.
- padding mask는 `[PAD]`를, causal mask는 미래 토큰을 보지 못하게 합니다.
- Self-Attention은 각 토큰이 이전 토큰의 정보를 어느 정도 결합할지 계산합니다.
- Attention 가중치만으로 모델의 언어 이해를 단정할 수는 없습니다.
