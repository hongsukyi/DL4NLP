# Lab 3 · 6개 모델 한 번에 비교 (MLP·Conv1D·RNN 계열)

**이 노트북은 한 세션에서 끝까지 이어서 실행합니다.** `res.json` 같은 파일을 주고받지 않고, 이 노트북
하나 안에서 6개 모델(Flatten/GAP/Conv1D/SimpleRNN/LSTM/GRU)을 전부 정의 → 학습 → 비교합니다.

- **먼저 6개 모델을 전부 정의**하고, **그다음 6개를 이어서 학습**합니다.
- 모델마다 하이퍼파라미터가 따로 있어서, 특정 모델만 튜닝하고 싶으면 그 변수만 바꾸면 됩니다.

> 6개를 전부 학습하므로 시간이 걸립니다. 중간에 끊기지 않도록 위에서 아래로 한 번에 실행하세요.

```
!pip install -q koreanize-matplotlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import koreanize_matplotlib

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.metrics import confusion_matrix
from sklearn.manifold import TSNE

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

## 1. 저장된 파일 불러오기

```
import urllib.request

url = "https://raw.githubusercontent.com/hongsukyi/DL4NLP/main/notebooks/data.npz"
urllib.request.urlretrieve(url, "data.npz")
```

```
data = np.load("data.npz", allow_pickle=True)

ids  = data["input_ids"]
mask = data["attention_mask"]
y    = data["labels"]

idx_tr  = data["train_idx"]
idx_val = data["val_idx"]
idx_te  = data["test_idx"]

vocab_size = int(data["vocab_size"])
max_len    = int(data["max_len"])
n_cls      = int(data["num_classes"])
classes    = data["class_names"].tolist()

X_tr,  M_tr,  y_tr  = ids[idx_tr],  mask[idx_tr],  y[idx_tr]
X_val, M_val, y_val = ids[idx_val], mask[idx_val], y[idx_val]
X_te,  M_te,  y_te  = ids[idx_te],  mask[idx_te],  y[idx_te]

print("train/val/test:", len(X_tr), len(X_val), len(X_te))
print("classes:", classes)
```

## 2. 공통 준비물 + 모델별 하이퍼파라미터

`dim`(임베딩 크기)만 6개 모델 공통이고, 나머지는 **모델마다 따로** 갖습니다.

```
res = {}  # 6개 모델 결과 (파일 저장 없이 메모리에만)

dim = 64  # 임베딩 크기 (공통)

# ---- 모델별 하이퍼파라미터 ----
flat_dense   = 64  # Flatten 은닉층 크기
gap_dense    = 64  # GAP 은닉층 크기
conv_filters = 64  # Conv1D 필터 수
conv_kernel  = 3   # Conv1D 커널 크기
rnn_units    = 64  # SimpleRNN 은닉상태 크기
lstm_units   = 64  # LSTM 은닉상태 크기
gru_units    = 64  # GRU 은닉상태 크기

def embed(inp, inm):
    """MLP 계열(Flatten/GAP/Conv1D)용: PAD 자리를 0으로 만든 임베딩"""
    x = layers.Embedding(vocab_size, dim)(inp)
    mf = layers.Lambda(lambda t: tf.cast(t, tf.float32)[..., None])(inm)
    return layers.Multiply()([x, mf])

def embed_mask(inp, inm):
    """RNN 계열(SimpleRNN/LSTM/GRU)용: PAD 자리를 0으로 만들고 Masking까지 적용"""
    x = embed(inp, inm)
    return layers.Masking(mask_value=0.0)(x)

def fit_model(model, name):
    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    es = keras.callbacks.EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)
    h = model.fit(
        [X_tr, M_tr], y_tr, validation_data=([X_val, M_val], y_val),
        batch_size=32, epochs=20, callbacks=[es], verbose=2,
    )
    return h

def show_result(model, h, name):
    fig, ax = plt.subplots(1, 2, figsize=(10, 3))
    ax[0].plot(h.history["loss"], label="train"); ax[0].plot(h.history["val_loss"], label="val")
    ax[0].set_title(f"{name} - Loss"); ax[0].legend()
    ax[1].plot(h.history["accuracy"], label="train"); ax[1].plot(h.history["val_accuracy"], label="val")
    ax[1].set_title(f"{name} - Accuracy"); ax[1].legend()
    plt.tight_layout(); plt.show()

    pred = model.predict([X_te, M_te], verbose=0).argmax(axis=1)
    acc = (pred == y_te).mean()
    res[name] = {
        "loss": h.history["loss"], "val_loss": h.history["val_loss"],
        "acc": h.history["accuracy"], "val_acc": h.history["val_accuracy"],
        "pred": pred.tolist(), "test_acc": float(acc),
    }
    print(f"[{name}] Test Accuracy: {acc:.3f}")

    cm = confusion_matrix(y_te, pred)
    plt.figure(figsize=(4, 4))
    plt.imshow(cm, cmap="Blues")
    plt.xticks(range(n_cls), classes); plt.yticks(range(n_cls), classes)
    for i in range(n_cls):
        for j in range(n_cls):
            plt.text(j, i, cm[i, j], ha="center", va="center")
    plt.title(f"{name} - Confusion Matrix"); plt.xlabel("예측"); plt.ylabel("정답")
    plt.tight_layout(); plt.show()

    feat_model = keras.Model(model.input, model.get_layer("feat").output)
    feat = feat_model.predict([X_te, M_te], verbose=0)
    xy = TSNE(n_components=2, random_state=SEED, init="pca", perplexity=30, n_jobs=1).fit_transform(feat)

    plt.figure(figsize=(5, 5))
    for c in range(n_cls):
        m = y_te == c
        plt.scatter(xy[m, 0], xy[m, 1], s=12, label=classes[c])
    plt.title(f"{name} - t-SNE"); plt.legend()
    plt.tight_layout(); plt.show()
```

## 3. 여섯 모델 정의하기 (아직 학습 안 함)

```
def build_flatten():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed(inp, inm)
    x = layers.Flatten()(x)
    x = layers.Dense(flat_dense, activation="relu")(x)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

flat_model = build_flatten()
```

```
def build_gap():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed(inp, inm)
    s = layers.Lambda(lambda t: tf.reduce_sum(t, axis=1))(x)
    c = layers.Lambda(lambda m: tf.maximum(
            tf.reduce_sum(tf.cast(m, tf.float32), axis=1, keepdims=True), 1.0))(inm)
    avg = layers.Lambda(lambda t: t[0] / t[1])([s, c])

    x = layers.Dense(gap_dense, activation="relu")(avg)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

gap_model = build_gap()
```

```
def build_conv1d():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed(inp, inm)
    x = layers.Conv1D(conv_filters, kernel_size=conv_kernel, padding="same", activation="relu")(x)

    bias = layers.Lambda(lambda m: (1.0 - tf.cast(m, tf.float32))[..., None] * -1e9)(inm)
    x = layers.Add()([x, bias])
    x = layers.GlobalMaxPooling1D()(x)

    x = layers.Dense(64, activation="relu")(x)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

conv_model = build_conv1d()
```

```
def build_simplernn():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed_mask(inp, inm)
    x = layers.SimpleRNN(rnn_units)(x)
    x = layers.Dense(64, activation="relu")(x)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

rnn_model = build_simplernn()
```

```
def build_lstm():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed_mask(inp, inm)
    x = layers.LSTM(lstm_units)(x)
    x = layers.Dense(64, activation="relu")(x)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

lstm_model = build_lstm()
```

```
def build_gru():
    inp = keras.Input((max_len,))
    inm = keras.Input((max_len,))

    x = embed_mask(inp, inm)
    x = layers.GRU(gru_units)(x)
    x = layers.Dense(64, activation="relu")(x)
    feat = layers.Dense(16, activation="relu", name="feat")(x)
    out = layers.Dense(n_cls, activation="softmax")(feat)

    return keras.Model([inp, inm], out)

gru_model = build_gru()
```

## 4. 여섯 모델 이어서 학습하기

정의는 끝났으니 `model.fit()`을 여섯 번 이어서 실행합니다. 특정 모델만 튜닝하고 싶으면 2절의
하이퍼파라미터(예: `lstm_units`)만 바꾸고 3절의 해당 모델 정의부터 다시 실행하면 됩니다.

```
h_flat = fit_model(flat_model, "Flatten")
```

```
h_gap = fit_model(gap_model, "GAP")
```

```
h_conv = fit_model(conv_model, "Conv1D")
```

```
h_rnn = fit_model(rnn_model, "SimpleRNN")
```

```
h_lstm = fit_model(lstm_model, "LSTM")
```

```
h_gru = fit_model(gru_model, "GRU")
```

## 5. 결과 확인

```
show_result(flat_model, h_flat, "Flatten")
```

```
show_result(gap_model, h_gap, "GAP")
```

```
show_result(conv_model, h_conv, "Conv1D")
```

```
show_result(rnn_model, h_rnn, "SimpleRNN")
```

```
show_result(lstm_model, h_lstm, "LSTM")
```

```
show_result(gru_model, h_gru, "GRU")
```

## 6. 6개 모델 성능 비교

```
names = list(res.keys())

table = pd.DataFrame({
    "모델": names,
    "Test Accuracy": [res[n]["test_acc"] for n in names],
}).set_index("모델")

print(table.round(3))
```

```
plt.figure(figsize=(7, 4))
plt.bar(table.index, table["Test Accuracy"])
plt.ylim(0, 1)
plt.title("모델별 테스트 정확도")
plt.ylabel("Accuracy")
for i, v in enumerate(table["Test Accuracy"]):
    plt.text(i, v + 0.01, f"{v:.3f}", ha="center")
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(7, 4))
for n in names:
    plt.plot(res[n]["val_loss"], marker="o", markersize=3, label=n)
plt.title("모델별 val_loss")
plt.xlabel("Epoch"); plt.ylabel("val_loss")
plt.legend()
plt.tight_layout()
plt.show()
```

```
best = table["Test Accuracy"].idxmax()
print(f"가장 정확도가 높은 모델: {best} ({table.loc[best, 'Test Accuracy']:.3f})")
```

## 정리

| 모델 | 하이퍼파라미터 |
|---|---|
| Flatten | `flat_dense` |
| GAP | `gap_dense` |
| Conv1D | `conv_filters`, `conv_kernel` |
| SimpleRNN | `rnn_units` |
| LSTM | `lstm_units` |
| GRU | `gru_units` |

`res.json` 같은 중간 파일 없이, 이 노트북 하나에서 6개 모델을 정의 → 학습 → 비교까지 끝냈습니다.
모델별 하이퍼파라미터가 분리되어 있어서, 특정 모델의 성능을 바꿔보고 싶으면 2절의 해당 변수만
고치고 그 모델의 정의·학습 셀만 다시 실행하면 됩니다.
